```python
import os
import re
import string
from flask import Flask, render_template
import nltk  # Import NLTK utama
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

# ==========================================================
# BAGIAN PERBAIKAN: Unduh data NLTK yang diperlukan
# ==========================================================
# Ini akan mengunduh paket 'stopwords', 'punkt', dan 'punkt_tab'
# jika belum ada di sistem.

try:
    nltk.data.find('corpora/stopwords')
    print("Resource 'stopwords' sudah ada.")
except LookupError:
    print("Resource 'stopwords' tidak ditemukan, mengunduh...")
    nltk.download('stopwords')

try:
    nltk.data.find('tokenizers/punkt')
    print("Resource 'punkt' sudah ada.")
except LookupError:
    print("Resource 'punkt' tidak ditemukan, mengunduh...")
    nltk.download('punkt')

# --- TAMBAHAN BARU UNTUK ERROR ANDA ---
try:
    nltk.data.find('tokenizers/punkt_tab')
    print("Resource 'punkt_tab' sudah ada.")
except LookupError:
    print("Resource 'punkt_tab' tidak ditemukan, mengunduh...")
    nltk.download('punkt_tab')
# --- AKHIR TAMBAHAN BARU ---

# ==========================================================
# AKHIR BAGIAN PERBAIKAN
# ==========================================================


# --- Inisialisasi Aplikasi Flask ---
app = Flask(__name__)

# --- Inisialisasi Preprocessing Tools (sesuai gambar) ---
# 1. Setup Stopwords (Bahasa Indonesia)
stop_words = set(stopwords.words('indonesian'))

# 2. Setup Stemmer (Sastrawi untuk Bahasa Indonesia)
factory = StemmerFactory()
stemmer = factory.create_stemmer()

# --- Fungsi Utama ---

def preprocess_text(text):
    """
    Fungsi untuk melakukan semua langkah preprocessing dari gambar.
    """
    # 1. Normalization / Lowercasing
    text = text.lower()
    
    # 2. Punctuation Removal
    # Menghapus angka dan tanda baca
    text = re.sub(r'\d+', '', text) # Hapus angka
    text = text.translate(str.maketrans('', '', string.punctuation)) # Hapus tanda baca
    text = text.strip() # Hapus spasi berlebih
    
    # 3. Segmentation / Tokenization
    tokens = word_tokenize(text)
    
    processed_tokens = []
    for token in tokens:
        # 4. Stopword Removal
        if token not in stop_words:
            # 5. Stemming (Lemmatization lebih rumit, kita pakai Stemming)
            stemmed_token = stemmer.stem(token)
            processed_tokens.append(stemmed_token)
            
    return processed_tokens

def create_inverted_index(corpus_path):
    """
    Fungsi untuk membaca dataset dan membangun Inverted Index.
    """
    inverted_index = {} # Dictionary untuk menampung index
    documents = {}      # Dictionary untuk menampung teks asli dokumen
    
    # Membaca semua file dari folder dataset
    for filename in os.listdir(corpus_path):
        if filename.endswith(".txt"):
            file_path = os.path.join(corpus_path, filename)
            
            with open(file_path, 'r', encoding='utf-8') as f:
                text = f.read()
                documents[filename] = text # Simpan teks asli
                
                # Lakukan preprocessing
                tokens = preprocess_text(text)
                
                # --- Feature Extraction (Inverted Index) ---
                unique_tokens = set(tokens)
                
                for token in unique_tokens:
                    # Jika token belum ada di index, buat list kosong
                    if token not in inverted_index:
                        inverted_index[token] = []
                    
                    # Tambahkan nama file (dokumen) ke posting list
                    inverted_index[token].append(filename)
    
    # Urutkan index berdasarkan abjad (term)
    sorted_index = dict(sorted(inverted_index.items()))
    
    return sorted_index, documents

# --- Rute Web (Flask) ---

@app.route('/')
def home():
    """
    Halaman utama web yang akan dipanggil saat user membuka web.
    """
    # Tentukan path ke folder dataset
    DATASET_FOLDER = 'dataset'
    
    # Buat inverted index dan ambil data dokumen
    index, docs = create_inverted_index(DATASET_FOLDER)
    
    # Kirim data (index dan docs) ke file HTML untuk ditampilkan
    return render_template('index.html', inverted_index=index, documents=docs)

# --- Menjalankan Aplikasi ---
if __name__ == '__main__':
    app.run(debug=True)
```
